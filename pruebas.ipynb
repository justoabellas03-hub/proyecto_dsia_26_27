```python
import json

import pandas as pd
import requests
pd.set_option("display.max_columns", None)

URL = "https://sedeaplicaciones.minetur.gob.es/ServiciosRESTCarburantes/PreciosCarburantes/EstacionesTerrestres/"

respuesta = requests.get(URL, headers={"Accept": "application/json"}, timeout=60)
respuesta.raise_for_status()
datos = json.loads(respuesta.content.decode("utf-8-sig"))

df = pd.DataFrame(datos["ListaEESSPrecio"])

df2=df.copy()
print(datos["Fecha"])   # momento al que corresponden los precios
print(df.shape)
df.head()
```

```python
def normalizar_numeros(df: pd.DataFrame) -> pd.DataFrame:

    resultado = df.copy()

    columnas_numericas = [
        col for col in resultado.columns
        if col.startswith("Precio") or col.startswith("%") or col in ("Latitud", "Longitud (WGS84)")
    ]

    for col in columnas_numericas:
        resultado[col] = resultado[col].str.replace(",", ".", regex=False)

    return resultado
```
